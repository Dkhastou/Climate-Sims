## Problem 1: Time-dependent warming in the zero-dimensional Energy Balance Model

### The equation of the Energy Balance Model

In lecture we defined a zero-dimensional Energy Balance Model for the global mean surface temperature $T_s$ as follows

$$ C  \frac{dT_s}{dt} = \text{ASR} - \text{OLR}$$

$$ \text{ASR} = (1-\alpha) Q $$

$$ \text{OLR} = \tau \sigma T_s^4$$

where we defined these terms:

- $C$ is a heat capacity for the atmosphere-ocean column
- $\alpha$ is the global mean planetary albedo
- $\sigma = 5.67 \times 10^{-8}$ W m⁻² K⁻⁴ is the Stefan-Boltzmann constant
- $\tau$ is our transmissivity parameter for the atmosphere.
- $Q$ is the global-mean incoming solar radiation, or *insolation*.

Refer back to our class notes for parameter values.

### Part (a)

If the heat penetrated to twice as deep into the ocean, the value of $C$ would be twice as large. Would this affect the **equilibrium temperature**? Why or why not? 

_Hint: look carefully at the equation for the Energy Balance Model. What is the condition for equilibrium? What role does $C$ play in the equation at equilibrium?_

#### Answer to Part (a)

Equlibrium temperature is the temperature when ASR and OLR are equal to each other:
1. The derivative is equal to the difference between the two
2. If the difference is 0, then the derivative is 0
3. If the derivative is 0, then we know the temperature is not changing
4. If the temperature is not changing, then the temperature is equilibrium temperature

But $C$ does not effect the difference between the two. Its only a proportionality constant of the derivative. We can see this more clearly in the fact that $C$ is not in the equations for either the ASR or the OLR, meaning that it can't effect either one. And since the difference is a function of both the ASR and OLR, if $C$ is not in the functions for either the ASR or OLR, the difference cannot be a function of $C$ either.

We can see this in the following graphs as well.

### Part (b)

In the [Lecture notes we used numerical timestepping](zero-dim-ebm#automate-the-timestepping-with-a-loop) to investigate a *hypothetical climate change scenario* in which $\tau$ decreases to 0.57 and $\alpha$ increases to 0.32. We produced a graph of $T_s(t)$ over a twenty year period, starting from an initial temperature of 288 K. Here you will repeat this calculate with a larger value of $C$ and compare the warming rates. Specifically:

- Repeat our in-class time-stepping calculation with the same parameters we used before (including a heat capacity of $C = 4\times10^8$ J m⁻² K⁻¹), but extend it to 50 years. **You should create an array of temperatures with 51 elements, beginning from 288 K**.
- Now do it again, but use $C = 8\times10^8$ J m⁻² K⁻¹ (representing 200 meters of water). You should **create another 51-element array** of temperatures also beginning from 288 K.
- **Make a well-labeled graph** that compares the two temperatures over the 50-year period.

```python
%matplotlib inline  
import numpy as np
import matplotlib.pyplot as plt

# Define Constants
sigma = 5.67E-8  # S-B constant
Tsobserved = 288.  # global average surface temperature
Q = 341.3  # the insolation

# Other Values
Freflected = 101.9  #  reflected shortwave flux in W/m2
alpha = Freflected / Q # The albedo
tau = 0.57
dt = 60. * 60. * 24. * 365.   # one year expressed in seconds
```

```python
numsteps = 50

def ASR(Q=Q, alpha=alpha):
    return (1-alpha)*Q

def OLR(T, tau=tau):
    return tau * sigma * T**4

def step_forward(T, C):
    return T + dt / C * ( ASR(alpha=0.32) - OLR(T, tau=0.57) )

# Reusable function to get data and plot it
def make_data(numsteps, C, label):
    for n in range(numsteps):
        Years[n+1] = n+1
        Tsteps[n+1] = step_forward(Tsteps[n], C)
    plt.plot(Years, Tsteps, label=label)

# Create Arrays
Tsteps = np.zeros(numsteps+1)
Years = np.zeros(numsteps+1)
Tsteps[0] = 288. 

make_data(numsteps, 4e8, r"$C = 4 x 10^8$")
make_data(numsteps, 8e8, r"$C = 8 x 10^8$" )

# Display
plt.title("Change in Temperature Over Time With Different Heat Capacities")
plt.xlabel("Time (years)")
plt.ylabel("Global mean temperature (Kelvin)")
plt.xlim(0, 50)
plt.ylim(288, 291.5)
plt.grid(True)
plt.legend()
plt.show()
```

## Problem 2: Albedo feedback in the Energy Balance Model

### The albedo formula

For this exercise, we will introduce a new physical process into our model by **letting the planetary albedo depend on temperature**. The idea is that a warmer planet has less ice and snow at the surface, and thus a lower planetary albedo (less reflection).

Represent the ice-albedo feedback through the following formula:

$$ \alpha(T) = \left\{ \begin{array}{ccc}
\alpha_i &   & T \le T_i \\
\alpha_o + (\alpha_i-\alpha_o) \frac{(T-T_o)^2}{(T_i-T_o)^2} &   & T_i < T < T_o \\
\alpha_o &   & T \ge T_o \end{array} \right\}$$

with the following parameter values:

- $\alpha_o = 0.289$ is the albedo of a warm, ice-free planet
- $\alpha_i = 0.7$ is the albedo of a very cold, completely ice-covered planet
- $T_o = 293$ K is the threshold temperature above which we assume the planet is ice-free
- $T_i = 260$ K is the threshold temperature below which we assume the planet is completely ice covered. 

For intermediate temperature, this formula gives a smooth variation in albedo with global mean temperature. It is tuned to reproduce the observed albedo $\alpha = 0.299$ for $T = 288$ K. 

### Part (a)

- Define a Python function that implements the above albedo formula. Your function should take the temperature $T$ as input and return the correct albedo $\alpha$. 
-  Use your function to calculate albedos for a wide range on planetary temperature (e.g. from $T=250$ K to $T=300$ K.)
- Present your results (albedo as a function of global mean temperature, or $\alpha(T)$) in a nicely labeled graph with temperature $T$ along the x axis and albedo $\alpha$ on the y axis.

_Hint: your code will need to decide which of the three formulas to use based on the value of $T$. You might want to review the use of `if` and `else` statements in Python. Alternatively, you can check out the `numpy.where()` function (see [the numpy documentation](https://numpy.org/doc/stable/reference/generated/numpy.where.html)) which can make decisions simultaneously on arrays of numbers. There is definitely more than one way to solve this problem. It doesn't matter how you do it as long as it works (and you understand why it works)!_

```python
# This is just the given function for albedo written in code
def albedo_function(T):
    alpha_o = 0.289
    alpha_i = 0.7
    T_o = 293 # In Kelvin
    T_i = 260 # In Kelvin
    
    if (T <= T_i):
        return alpha_i
    elif (T >= T_o):
        return alpha_o
    elif ((T_i < T) and (T < T_o)):
        return ( alpha_o + ( (alpha_i - alpha_o) * ( (T-T_o)**2 / (T_i-T_o)**2) ) )

# Temperature starts at 250 to graphically show the bounds of the function for a given input T
temp = 250
numsteps = 100

# Make arrays
Temperatures = np.zeros(numsteps)
Albedo = np.zeros(numsteps)

# Iterate through arrays
for i in range(numsteps):
    Temperatures[i] = temp
    Albedo[i] = albedo_function(temp)
    temp = temp + 0.5

# Plot and display
plt.plot(Temperatures, Albedo)
plt.title(r"Albedo vs Temperature")    
plt.grid(True)
plt.xlim(250, 300)
plt.xlabel("Temperature (Kelvin)")
plt.ylabel("Albedo")
plt.show()
```

For this graph, I tried to find a good balance between the number of steps used and the temperature increments. The behavior here is interesting. First of all, it seems unphysical that the graph would begin decreasing sharply at $T = 260 K$ (The point is undifferentiable, and in physics we would typically assume smoothness everywhere, so the model is probably too simplistic here).

Besides this, the implications this has for what we can expect from climate change don't seem to be in our favor. It would be nicer if the graph decreased more gradually and linearly, giving us a much more controlled descent. We instead see the albedo drop off sharply and suddenly as temperature increases, and the descent only slows down once the temperature gets much hotter 

### Part (b)

Now investigate a climate change scenario with this new model:

- Suppose that the transmissivity decreases from 0.611 to 0.57 (same as before)
- This time the albedo is not a fixed, known number. Instead, it varies with the current value of temperature according the formula you plotted in Part (a).
- Your task is to **calculate the new equilibrium temperature**. 
- Unlike when the albedo is a fixed number, it is not possible to solve for the equilibrium temperature analytically. Instead, you will use numerical time-stepping to find the equilibrium temperature
- Repeat the procedure from Problem 1b *(time-step forward for 50 years from an initial temperature of 288 K and make a graph of the results)*, but this time **use the function you defined above to compute the albedo for the current temperature**.

```python
# Reset old values
numsteps = 75
C = 4e8

def ASR(alpha=alpha):
    return (1-alpha)*Q

def OLR(T, tau=tau):
    return tau * sigma * T**4

# Mostly the same code as before, but here we add a condition to determine if we should use the function or not
# to compare graphs. If yes, we set alpha equal to the output of the albedo function
def step_forward(T, albedo, use_function):
    if use_function == False:
        return T + dt / C * ( ASR(alpha=alpha) - OLR(T, tau=0.57) )
    if use_function == True:
        return T + dt / C * ( ASR(alpha=albedo) - OLR(T, tau=0.57) )
        
def make_data(use_function, label):
    Tsteps[0] = 269.25 # Define initial Temperature
    for n in range(75):
        albedo = albedo_function(Tsteps[n])
        Years[n+1] = n+1
        Tsteps[n+1] = step_forward(Tsteps[n], albedo, use_function)
    plt.plot(Years, Tsteps, label=label)
    print(f"Final temperature for {label}:", Tsteps[numsteps])

# Create Arrays
Tsteps = np.zeros(numsteps+1)
Years = np.zeros(numsteps+1)

make_data(False, "Constant albedo") # The original graph, no change
make_data(True, "Changing albedo") # Now decreasing


# Display
plt.title("Change in Temperature Over Time With Different Albedo Functions")
plt.xlabel("Time (years)")
plt.ylabel("Global Mean Temperature (Kelvin)")
plt.xlim(0, 75)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Q = 341.3  # The original insolation

numsteps = 75
C = 4e8

def ASR(Q, alpha=alpha):
    return (1-alpha)*Q

def OLR(T, tau=tau):
    return tau * sigma * T**4

def step_forward(T, albedo, Q):
    return T + dt / C * ( ASR(Q, alpha=albedo) - OLR(T, tau=0.57) )
        
def make_data(Q, label):
    Tsteps[0] = 269.25 # Define initial Temperature
    for n in range(75):
        albedo = albedo_function(Tsteps[n])
        Years[n+1] = n+1
        Tsteps[n+1] = step_forward(Tsteps[n], albedo, Q)
    plt.plot(Years, Tsteps, label=label)

# Create Arrays
Tsteps = np.zeros(numsteps+1)
Years = np.zeros(numsteps+1)

make_data(341.3, r"$Q = 341.3$") # The original graph, no change
make_data(341.1, r"$Q = 341.1$") # Now decreasing
make_data(340.9, r"$Q = 340.9$")
make_data(340.7, r"$Q = 340.7$")

# Display
plt.title("Change in temperature over time with different insolation values")
plt.xlabel("Time (years)")
plt.ylabel("Global Mean Temperature (Kelvin)")
plt.xlim(0, 75)
plt.grid(True)
plt.legend()
plt.show()
```

#### Thoughts on Decreasing $Q$

I actually thought I did something wrong when I did this (and who knows, maybe I did). But looking at the y axis, we see that the temperature is still starting from the same point. Which means that subtly decreasing the insolation actually causes the planet to cool. We could maybe say that in the giant configuration space of the model, the point that the insolation value we've been using lies at is unstable - slightly perturbing it causes a big difference in the outcome of the global mean temperature over 50 years

Its also important to think about if this actually makes sense physically. Insolation is the global mean incoming solar radiation, and as per the equation above, we have 

$$ \text{ASR} = (1-\alpha) Q $$

From this equation, we can see that decreasing $Q$ will decrease the ASR. Then from the main equation, after solving for $\frac{dT_s}{dt}$ we have

$$ \frac{dT_s}{dt} = (\text{ASR} - \text{OLR})/C$$

From our graphs, we see that the rate of change of the temperature goes from increasing to decreasing as insolation decreases. This means that the derivative must be changing from a positive value to a negative value as the ASR decreases. And the only way for the derivative to become negative is if the OLR becomes larger than the ASR. So this makes sense mathematically. Physically, whats happening is that as the insolation drops, the radiation leaving the planet becomes larger than the radiation being absorbed by the planet, and this causes the planet to cool over time.

#### Increasing transmitivitty

Now we will try to increase transmitivitty $\tau$ while returning $Q$ to its original constant value

The way the bonus problem is worded makes it sound like we should observe a similar, if not exactly the same trend, by increasing transmitivitty. From the equation for OLR:

$$ \text{OLR} = \tau \sigma T_s^4$$

We can see that this would make sense, because increasing $\tau$ should increase the OLR, causing it to become higher than the ASR, which should cause the graph to invert in the same way as before.

```python
# We will use basically the same exact code but just change the input to our be tau instead of Q

Q = 341.3  # The insolation
# tau = 0.57 Original transmitivitty value

numsteps = 75
C = 4e8

def ASR(Q, alpha=alpha):
    return (1-alpha)*Q

def OLR(T, tau=tau):
    return tau * sigma * T**4

def step_forward(T, albedo, tau):
    return T + dt / C * ( ASR(Q, alpha=albedo) - OLR(T, tau=tau) )
        
def make_data(tau, label):
    Tsteps[0] = 269.25 # Define initial Temperature
    for n in range(75):
        albedo = albedo_function(Tsteps[n])
        Years[n+1] = n+1
        Tsteps[n+1] = step_forward(Tsteps[n], albedo, tau)
    plt.plot(Years, Tsteps, label=label)

# Create Arrays
Tsteps = np.zeros(numsteps+1)
Years = np.zeros(numsteps+1)

make_data(0.57, r" $\tau = 0.57$") # The original graph, no change
make_data(0.58, r" $\tau = 0.58$") # Now decreasing
make_data(0.59, r" $\tau = 0.59$")
make_data(0.60, r" $\tau = 0.60$")

# Display
plt.title("Change in Temperature Over Time With Different Transmitivitty Values")
plt.xlabel("Time (years)")
plt.ylabel("Global Mean Temperature (Kelvin)")
plt.xlim(0, 75)
plt.grid(True)
plt.legend()
plt.show()
```

#### Thoughts on increasing $\tau$ 

From these graphs, we observe basically the same thing, which is expected. There are some differences though:
1. The model actually seems even more sensitive to $\tau$ being changed than to $Q$ being changed. Why? Probably because in our equation $$ \text{OLR} = \tau \sigma T_s^4$$
   $\tau$ is a multiple of $T_s^4$ which means that changing it will change the OLR much more, and therefore change the difference more
2. Unlike the equlibrium reached when we perturb $Q$, perturbing $\tau$ causes different equilibrium temperatures after cooling/warming
