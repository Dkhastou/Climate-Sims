### Step 1: Setting up our reference model

````python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
from metpy.plots import SkewT
import climlab
````

````python
#  This code is used just to create the skew-T plot of global, annual mean air temperature
time_coder = xr.coders.CFDatetimeCoder(use_cftime=True)

# We usually default to the data served from NOAA ESRL
ncep_url = "http://www.esrl.noaa.gov/psd/thredds/dodsC/Datasets/ncep.reanalysis.derived/"
ncep_air = xr.open_dataset( ncep_url + "pressure/air.mon.1981-2010.ltm.nc", decode_times=time_coder)

# Alternate data source if the NOAA ESRL server is down (not in use now)

# url = 'http://apdrc.soest.hawaii.edu:80/dods/public_data/Reanalysis_Data/NCEP/NCEP/clima/pressure/air'
# THE LINE BELOW IS BUGGED, SOME ISSUE WITH xarray I DON'T WANT TO BOTHER WITH
# air = xr.open_dataset(url)
# The name of the vertical axis is different than the NOAA ESRL version..
# ncep_air = air.rename({'lev': 'level'})

# Take global, annual average 
coslat = np.cos(np.deg2rad(ncep_air.lat))
weight = coslat / coslat.mean(dim='lat')
Tglobal = (ncep_air.air * weight).mean(dim=('lat','lon','time'))

#  Resuable function to plot the temperature data on a Skew-T chart
def make_skewT():
    fig = plt.figure(figsize=(9, 9))
    skew = SkewT(fig, rotation=30)
    skew.plot(Tglobal.level, Tglobal, color='black', linestyle='-', linewidth=2, label='Observations')
    skew.ax.set_ylim(1050, 10)
    skew.ax.set_xlim(-90, 45)
    # Add the relevant special lines
    skew.plot_dry_adiabats(linewidth=0.5)
    skew.plot_moist_adiabats(linewidth=0.5)
    #skew.plot_mixing_lines()
    skew.ax.legend()
    skew.ax.set_xlabel('Temperature (degC)', fontsize=14)
    skew.ax.set_ylabel('Pressure (hPa)', fontsize=14)
    return skew

#  and a function to add extra profiles to this chart
def add_profile(skew, model, linestyle='-', color=None):
    line = skew.plot(model.lev, model.Tatm - climlab.constants.tempCtoK,
             label=model.name, linewidth=2)[0]
    skew.plot(1000, model.Ts - climlab.constants.tempCtoK, 'o', 
              markersize=8, color=line.get_color())
    skew.ax.legend()

# Get the water vapor data from CESM output
cesm_data_path = "http://thredds.atmos.albany.edu:8080/thredds/dodsC/CESMA/"
# For better performance if you can access the roselab_rit filesystem (e.g. from JupyterHub)
# cesm_data_path = "/roselab_rit/cesm_archive/"
atm_control = xr.open_dataset(cesm_data_path + "cpl_1850_f19/concatenated/cpl_1850_f19.cam.h0.nc")
# Take global, annual average of the specific humidity
weight_factor = atm_control.gw / atm_control.gw.mean(dim='lat')
Qglobal = (atm_control.Q * weight_factor).mean(dim=('lat','lon','time'))
````

````python
#  Make a model on same vertical domain as the GCM
mystate = climlab.column_state(lev=Qglobal.lev, water_depth=2.5)
#  Build the radiation model -- just like we already did
rad = climlab.radiation.RRTMG(name='Radiation',
                              state=mystate, 
                              specific_humidity=Qglobal.values,
                              timestep = climlab.constants.seconds_per_day,
                              albedo = 0.25,  # surface albedo, tuned to give reasonable ASR for reference cloud-free model
                             )
#  Now create the convection model
conv = climlab.convection.ConvectiveAdjustment(name='Convection',
                                               state=mystate,
                                               adj_lapse_rate=6.5,
                                               timestep=rad.timestep,
                                              )
#  Here is where we build the model by coupling together the two components
rcm = climlab.couple([rad, conv], name='Radiative-Convective Model')
````

This rcm is our reference model which we will compare our more complex models to. We can find the amount of CO2 in this model using the line of code below.

````python
rcm.subprocess['Radiation'].absorber_vmr['CO2']
````

### Step 2: Run the model to equilibrium

We then "integrate" the model over 5 years, which should set it in equilibrium. 

````python
rcm.integrate_years(5)
````

````python
print(rcm.ASR - rcm.OLR)
````

Here we are finding the energy balance between the Absorbed shortwave radiation and outgoing longwave radiation. This is quite similar to what we did in class - the model has to be in equilibrium. So the difference between the two should be 0. And the code above shows that the difference between the two is so small that it is basically 0. Therefore we are in equilibrium. 

### Step 3: Make a new model with double the atmospheric CO2

````python
# Make an exact clone with same temperatures
rcm_2xCO2 = climlab.process_like(rcm)
rcm_2xCO2.name = 'Radiative-Convective Model (2xCO2 initial)'
````

````python
#  Check to see that we indeed have the same CO2 amount
rcm_2xCO2.subprocess['Radiation'].absorber_vmr['CO2']
````

````python
#  Now double it!
rcm_2xCO2.subprocess['Radiation'].absorber_vmr['CO2'] *= 2
````

````python
#  and verify
rcm_2xCO2.subprocess['Radiation'].absorber_vmr['CO2']
````

We then need to get the atmosphere-adjusted radiative forcing, given by the equation:

$$ \Delta R = \left(\text{ASR}_{2xCO2} - \text{OLR}_{2xCO2}\right) - \left(\text{ASR}_{ref} - \text{OLR}_{ref}\right)$$

### Step 3.1 Calculate the atmosphere-adjusted radiative forcing using our new model

````python
rcm_2xCO2.compute_diagnostics()
````

````python
print(rcm_2xCO2.ASR - rcm.ASR)
````

````python
print(rcm_2xCO2.OLR - rcm.OLR)
````

````python
DeltaR_adjusted = (rcm_2xCO2.ASR - rcm_2xCO2.OLR) - (rcm.ASR - rcm.OLR)
print(DeltaR_adjusted)
````

### Step 4: Make another model clone with doubled CO$_2$, and timestep the model out to equilibrium.

````python
# Make another copy, this time of our doubled CO2 model
rcm_2xCO2_adjusted = climlab.process_like(rcm_2xCO2)
rcm_2xCO2_adjusted.name = 'Radiative-Convective Model (2xCO2 atmosphere-adjusted)'

# Time step the model
for n in range(2000):
    rcm_2xCO2_adjusted.step_forward()
    # hold surface temperature fixed by setting it equal to our reference models surface temperature
    rcm_2xCO2_adjusted.Ts[:] = rcm.Ts[:]
````

````python
print(rcm_2xCO2_adjusted.ASR - rcm_2xCO2_adjusted.OLR)
````

From my understanding, the temperature is not in equilibrium here because we are holding the surface temperature constant and not letting it balance out.

````python
rcm_2xCO2_eq = climlab.process_like(rcm_2xCO2_adjusted)
rcm_2xCO2_eq.name = 'Radiative-Convective Model (2xCO2 equilibrium)'
rcm_2xCO2_eq.integrate_years(5)
````

````python
# are we close to equilibrium?
print(rcm_2xCO2_eq.ASR - rcm_2xCO2_eq.OLR)
````

Now our temperature is in equilibrium

````python
skew = make_skewT()
add_profile(skew, rcm)
add_profile(skew, rcm_2xCO2_adjusted)
add_profile(skew, rcm_2xCO2_eq)
````

### Some notes on the graph

This graph is showing us a couple of things. 

First, we can see that the observed data obeys a different pattern compared to our model, which is expected. Our model looks a little too 'simple', in that it follows much straighter lines and and doesn't curve as gradually near the ~200 hPa point on the graph.

Second, its a little hard to make out whats going on, but the atmosphere adjusted model and regular atmosphere model lines are basically overlapping near the bottom and are therefore the same. Only after we integrate the adjusted atmosphere model out over time can we see the difference in surface temperature between the model without double CO2 and the one with it. And we can see that the temperature at the surface as a result is hotter, which is also given via with the ECS in step 5 below.

### Step 5: Calculate the no-feedback Equilibrium Climate Sensitivity (ECS)

````python
ECS_nofeedback = rcm_2xCO2_eq.Ts - rcm.Ts
print(ECS_nofeedback)
````

Here we can see the precise difference in surface temperatures between the two models, which is about 1.3 K. 

### Step 6: Calculate the no-feedback climate response parameter $\lambda_0$ in W m$^{-2}$ K$^{-1}$.

````python
lambda0 = DeltaR_adjusted / ECS_nofeedback
print(lambda0)
````

    - Modify the specific humidity so that the relative humidity remains constant.

Verify and show that you get the same ECS for this version of the model including a water vapor feedback as we did in the lecture notes.

### Step 1: Calculate the relative humidity profile in the reference RCM

````python
#  actual specific humidity
q = rcm.subprocess['Radiation'].specific_humidity

#  saturation specific humidity (a function of temperature and pressure)
qsat = climlab.utils.thermo.qsat(rcm.Tatm, rcm.lev)

#  Relative humidity
rh = q/qsat
````

We calculate the relative humidity of our model using the actual specific humidity and the saturation specific humidity. This comes from the formula we used in class:

$$ RH = 100 \times \frac{q}{q_s}$$


### Steps 2 and 3: Make a new clone of the model with doubled CO$_2$ and run it out to equilibrium

````python
rcm_2xCO2_h2o = climlab.process_like(rcm_2xCO2)
rcm_2xCO2_h2o.name = 'Radiative-Convective Model (2xCO2 equilibrium with H2O feedback)'

for n in range(2000):
    # At every timestep
    # we calculate the new saturation specific humidity for the new temperature
    #  and change the water vapor in the radiation model
    #  so that relative humidity is always the same
    
    qsat = climlab.utils.thermo.qsat(rcm_2xCO2_h2o.Tatm, rcm_2xCO2_h2o.lev)
    rcm_2xCO2_h2o.subprocess['Radiation'].specific_humidity[:] = rh * qsat
    rcm_2xCO2_h2o.step_forward()
````

This part was genuinely annoying! You have to create the new model and run the model in the *same cell*, otherwise the model won't be reset each time you run it. (The lectures don't do this). This is one of the quirks with jupyter notebook and the reason I don't find it particularly nice to work with beyond the introductory level.

````python
# Check for balance again
print(rcm_2xCO2_h2o.ASR - rcm_2xCO2_h2o.OLR)
````

We check for energy balance and find it satisfactory. I originally found doing this step every time annoying, but as a reference to the point I made in the cell above, checking that we're getting expected values is really important otherwise we will be working with models that have been timestepped or configured incorrectly without realizing it. Checking for equlibrium balance is part of what helped me figure out how to troubleshoot some of these issues

### Step 4: Calculate the ECS for this new version of the model

````python
ECS = rcm_2xCO2_h2o.Ts - rcm.Ts
print(ECS)
````

It looks like the difference in surface temperatures for our h20 model is much greater than before. We can plot this on a skew-T again to see whats happening.

````python
# Because we already know that rcm and rcm_2xCO2_adjusted are basically the same, we will leave out rcm_2xCO2_adjusted

skew = make_skewT()
add_profile(skew, rcm)
# add_profile(skew, rcm_2xCO2_adjusted)
add_profile(skew, rcm_2xCO2_eq)
add_profile(skew, rcm_2xCO2_h2o)
````

## Question 3: combined lapse rate and water vapor feedback in the RCM

*Now you are going to do something new, but the process will be very similar to what you did in Question 2.*

### Instructions

A typical, expected feature of global warming is that the **upper troposphere warms more than the surface**. (Later we will see that this does occur in the CESM simulations).

This feature is **not represented in our standard radiative-convective model**, which is forced to a single prescribed lapse rate due to our convective adjustment.

Here you will suppose that other physical processes modify this lapse rate as the climate warms. 

**Repeat the RCM global warming calculation, but implement two different feedbacks:**

- a water vapor feedback using **fixed relative humidity** (i.e., what you did in Question 2)
- a **lapse rate feedback** using this formula:

$$ \Gamma = 6.5 \text{ K} \text{ km}^{-1} - (0.3 \text{ km}^{-1}) \Delta T_s $$

where 6.5 K km$^{-1}$ is the critical lapse rate you used in your control model, and $\Delta T_s$ is the **current value of the surface warming relative to the control** in units of K. 

So, for example if the model has warmed by 1 K at the surface, then our parameterization says that the critical lapse rate should be $6.5 - 0.3 \times 1 = 6.2$ K km$^{-1}$.

Follow the example in the lecture notes where we implemented the fixed relative humidity. Remember that we needed to adjust the `specific_humidity` field of our radiation process at every timestep. Here we still need to do that, but in addition, we also need to change the attribute

```
adj_lapse_rate
```

of the convection process at each timestep.

For example, if you have a model called `mymodel` that contains a `ConvectiveAdjustment` process called `Convection`, here is some code that you would need inside your `for` loop at every timestep:
```
mymodel.subprocess['Convection'].adj_lapse_rate = newvalue
```
where `newvalue` is a number in K / km. 

As in Question 2, you will need to write a `for` loop where, at every timestep, you not only set the required specific humidity, but you also calculate the appropriate current value for $\Gamma$ using the current surface warming, and set the `adj_lapse_rate` parameter using this value as shown above. 


### Specific questions:

1. Make a nice skew-T plot that shows three temperature profiles:
    - RCM control
    - RCM, equilibrium after doubling CO$_2$ without feedback
    - RCM, equilibrium after doubling CO$_2$ with combined water vapor and lapse rate feedback
2. Based on your plot, where in the column do you find the greatest warming?
3. Calculate the ECS of the new version of the model with combined water vapor and lapse rate feedback
4. Is this sensitivity larger or smaller than the "no feedback" ECS? Is it larger or smaller than the ECS with water vapor feedback alone (which we calculated in the lecture notes, and you verified in Question 2)?
5. Calculate the combined feedback parameter for (water vapor plus lapse rate).
6. Compare this result to the IPCC figure with feedback results from comprehensive models in our lecture notes (labeled "WV+LR"). Do you find a similar number?
7. Would you describe the **lapse rate feedback** as positive or negative? Explain why.

````python
rcm_2xCO2_lapserate = climlab.process_like(rcm_2xCO2)
rcm_2xCO2_lapserate.name = 'Radiative-Convective Model (2xCO2 equilibrium with H2O and lapse rate feedback)'

criticalLapseRate = 6.5

# Lapse rate function
def calculate_lapse_rate(Delta_Ts):
    Gamma = criticalLapseRate - (0.3 * Delta_Ts)
    return Gamma
    
for n in range(2000):
    
    # At every timestep we calculate the new saturation specific humidity for the new temperature
    # and change the water vapor in the radiation model so that relative humidity is always the same
    qsat = climlab.utils.thermo.qsat(rcm_2xCO2_lapserate.Tatm, rcm_2xCO2_lapserate.lev)
    rcm_2xCO2_lapserate.subprocess['Radiation'].specific_humidity[:] = rh * qsat

    # Get the Delta_Ts value
    Delta_Ts = rcm_2xCO2_lapserate.Ts - rcm.Ts

    # I had issues here with the kernel restarting over and over. I was using rcm_2xCO2_lapserate.Ts without subtracting
    # rcm.Ts for the Delta_Ts value which was probably massively overloading the model. Whats annoying is that I didn't
    # get a useful error message or any error message at all. (I talk about this more in the conclusion).
    
    # Feed the current temperature into the lapse rate formula using the Delta_Ts value
    rcm_2xCO2_lapserate.subprocess['Convection'].adj_lapse_rate = calculate_lapse_rate(Delta_Ts)

    # Step the model forward in time
    rcm_2xCO2_lapserate.step_forward()
````

````python
skew = make_skewT()
add_profile(skew, rcm)
# add_profile(skew, rcm_2xCO2_adjusted)
add_profile(skew, rcm_2xCO2_eq)
add_profile(skew, rcm_2xCO2_h2o)
add_profile(skew, rcm_2xCO2_lapserate)
````
